# Car Price Prediction Inference
This notebook loads a trained XGBoost model and predicts car prices based on user input features.

```python
import xgboost as xgb
import numpy as np
import pandas as pd
```

## Load the trained model
Load the XGBoost model from the provided path.

```python
model = xgb.XGBRegressor()
model.load_model('d:/Coding/Exploratory_Data_Analysis/models/provided_data_model.json')
```

## Enter Car Features
Input the car features below. Ensure the order and type matches the model's training data.

```python
# Example feature names (update as per your model):
feature_names = [
    'mileage','engine_cc','fuel','kilometer_run','is_automatic','is_4WD','age'
]

# Ask user for input (for demonstration, using input() for each feature)
user_input = []
for name in feature_names:
    value = input(f'Enter {name}: ')
    user_input.append(value)

# Convert to DataFrame and ensure numeric types where needed
input_df = pd.DataFrame([user_input], columns=feature_names)
for col in ['mileage','engine_cc','fuel','kilometer_run','is_automatic','is_4WD','age']:
    input_df[col] = pd.to_numeric(input_df[col], errors='coerce')
```

## Predict and Display Output
Use the loaded model to predict the car price and display the result.

```python
pred_log = model.predict(input_df)
pred_price = np.expm1(pred_log)[0]
print(f'Predicted Car Price: {pred_price:.2f}')
```
